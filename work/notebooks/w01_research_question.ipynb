## 2. The question: decision, action, cost of a wrong call

- **Research Question:** How can we prioritize declining and decay-risk content pages for review so content teams maximize traffic preservation per editorial hour spent?
- **Unit of Analysis (Grain):** A single pseudonymized content page (`content_id`).
- **Decision Improved:** Which top 20–50 candidate pages should an editorial team review and update first during their weekly refresh sprint?
- **Concrete Action:** Initiating a content refresh (updating stale statistics, rewriting meta titles/descriptions, expanding thin content, or fixing intent alignment).
- **Cost of a Wrong Call (False Positive / False Negative):**
  - *False Positive (Flagging a healthy page):* Wasted editorial hours spent auditing pages that do not actually need intervention, creating an opportunity cost.
  - *False Negative (Missing a declining page):* Unnoticed traffic decay that compound over months until recovering search rankings requires a full ground-up rewrite.
- **Why ML / Data Helps:** Static hand-rules (like `days_since_update > 180`) treat all pages equally and ignore non-linear interactions between search volume, position tier, CTR, and engagement rates. ML ranks pages probabilistically based on historical signals to surface high-impact candidates automatically.

```python
# Section 2: Framework Summary
print("Decision: Weekly refresh prioritization queue")
print("Unit of Analysis: Single content page (content_id)")
```

## 3. Quick look at the data (2-3 real numbers)

Below, we load `data/raw/content_refresh_anonymized.csv` live and extract 3 core baseline numbers that justify why this lane is worth our focus over the next 7 weeks:

1. **Total Dataset Volume & Overall Decline Rate:** Out of 30,000 pseudonymized pages, **16,262 pages (54.2%)** exhibit a downward traffic trend (`trend_direction == 'down'`).
2. **Visible Demand Exposure:** **22,006 pages (73.4%)** receive substantial demand (`impressions_90d >= 100`).
3. **High-Impact Refresh Candidates:** Among pages with active demand (`impressions_90d >= 100`), **13,152 pages (59.8%)** are actively declining — establishing a massive backlog that demands prioritized decision support.

```python
import csv, os

data_path = "data/raw/content_refresh_anonymized.csv"
if not os.path.exists(data_path):
    # Fallback search if kernel started elsewhere
    data_path = "../../data/raw/content_refresh_anonymized.csv"

with open(data_path, "r", encoding="utf-8") as f:
    reader = list(csv.DictReader(f))

total_pages = len(reader)
declining_pages = sum(1 for row in reader if row.get("trend_direction") == "down")
visible_pages = sum(1 for row in reader if float(row.get("impressions_90d", 0) or 0) >= 100)
declining_visible = sum(1 for row in reader if float(row.get("impressions_90d", 0) or 0) >= 100 and row.get("trend_direction") == "down")

print("=== LIVE DATASET ANALYSIS (data/raw/content_refresh_anonymized.csv) ===")
print(f"1. Total Pages Analyzed:      {total_pages:,}")
print(f"2. Overall Declining Pages:    {declining_pages:,} ({declining_pages/total_pages:.2%} of total inventory)")
print(f"3. Visible Pages (imp >= 100): {visible_pages:,} ({visible_pages/total_pages:.2%} of total inventory)")
print(f"4. Declining Visible Pages:    {declining_visible:,} ({declining_visible/visible_pages:.2%} of visible demand)")
print("\nConclusion: Over 13,000 visible pages show declining trends. ML-based ranking will isolate the top 50 highest-impact opportunities for human review.")
```

## 4. Careful words: what I can and can't claim

### What I CAN claim:
- **Observational Correlations:** Signals associated with higher decline rates across the historical snapshot.
- **Decision Support:** A probabilistic ranking model that orders candidate pages by decline risk better than random or simple hand-rules.
- **Directional Guidance:** Identifying specific features (e.g. content age, CTR relative to position tier) that correlate with traffic movement.

### What I CANNOT claim:
- **Causal Proof:** We cannot claim that updating a flagged page will *cause* its search traffic to recover (causal inference requires randomized control trials or experimental designs).
- **Reverse-Engineering Algorithms:** We are not "predicting Google's algorithm" or uncovering private ranking factors.
- **Deterministic Guarantees:** Model scores represent empirical probabilities on past observable data, not guaranteed future performance.
